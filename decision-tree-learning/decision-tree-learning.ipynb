Detta kan bero på att de kännetecken vi ändrade för de nya djuren kan ha varit särdrag som generellt gav mindre ny informationstillförsel för beslut och för att flera spalter inte ändrades. I detta test som gjordes verkade algoritmen robust mot felaktig data men vi har svårt att föreställa oss att detta även gäller för större mängder felaktig data där den felaktiga datan utgör en större andel av den totala datan. 

```python
import random as r

class MyDecisionTreeLearner(DecisionTreeLearner):
    
    def information_gain(self, attr, examples):
        def I(examples):
            target = self.dataset.target
            return information_content([self.count(target, v, examples)
                                        for v in self.dataset.values[target]])
        N = float(len(examples))
        remainder = 0
        for (v, examples_i) in self.split_by(attr, examples):
            remainder += (len(examples_i) / N) * I(examples_i)
        #return I(examples) - remainder
        return r.randint(0,1000)

train_and_test_all(MyDecisionTreeLearner(), zoo, 0, 30, print_tree=True)
```

**Svar**: 

Generella fördelar med beslutträdsinlärning är att stegen som tas för att nå ett beslut eller en klassificering är enkel att följa samt att visualisera. Däremot kan enkelheten snabbt bli för komplec då storleken på datan ökar. En ytterligare nackdel är dess känslighet för träningsdatan, vilket diskuteras vidare nedan.

Storleken på träningssetet är väsentligt för hur väl modellen generaliserar. Är det för litet blir modellen underoptimiserad för träningsdatan, vilket leder till att mönster som är relevanta inte upptäcks och precisionen blir lägre. Är det för stort blir modellen överoptimiserad för träningsdatan, vilket leder till att mönster möjligtvis som enbart är relevanta i träningsdatan blir centrala i modellen och därav blir precisionen lägre.

De aspekter som utmärker ett bra dataset är att datan är relevant, balanserad och har korrekt storlek för att undvika under- och överanpassning. Att ett dataset är relevant innebär att den bör innehålla exempel på objekt som är liknande de objekt som beslutsträdet sedan ska appliceras på. Är datasetet inte relevant kommer inte de mönster som existerar i datasetet existera för ny data. Att ett dataset är balanserat innebär att det har en jämn fördelning mellan olika "klasser". En ojämn fördelning skulle innebära en bias i modellen mot särdrag hos de överrepresentterade klasserna, vilket skulle leda till att objekt av andra klasser möjligtvis klassificeras baserat på egenskaper som är förhållandevis irrelevanta för just den klassen. Därav skulle modellens precision minska.

Att använda beslutträdsinlärning för större mängder data kan leda till problem. Främst att tidskomplexiteten ökar men även att modellen blir svår att förstå. Däremot finns där metoder för att effektivisera användning av beslutträdsmodeller för större mängder data. Dock uppstår även problem kopplade till de höga kraven på träningsdatan som används för inlärning. Alltså att datan behöver vara klassificerad och balanserad. Detta kan bidra med nackdelar mot att använda beslutträdsinlärning för större problem då arbetet med testdatan kan bli för stort. Givet alldeles för stora problem finns där möjligtvis mer effektiva metoder att använda men beslutträdsinlärning är definitivt en användbar metod vid klassificering och besultsfattande.
